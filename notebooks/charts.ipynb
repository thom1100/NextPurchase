```python
import sys
from pathlib import Path

scripts_path = Path("..") / "scripts"
sys.path.append(str(scripts_path))

from utils import *
import plotly.express as px
```

```python
tx = pd.read_csv("../data/transformed/tx.csv")
```

```python
sales_by_category = (
    tx.groupby("Category", as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"))
    .sort_values("revenue", ascending=False)
)

fig = px.bar(sales_by_category, x="Category", y="revenue",
             title="Revenue by category")
fig.update_layout(template="plotly_white", xaxis_tickangle=-45)
fig.show()
```

```python
sales_by_country = (
    tx.groupby("StoreCountry", as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"),
         units=("Quantity", "sum"))
    .sort_values("revenue", ascending=False)
)
sales_by_country["iso_alpha"] = sales_by_country["StoreCountry"].apply(country_to_iso3)
```

```python
fig = px.choropleth(
    sales_by_country,
    locations="iso_alpha",
    color="revenue",
    hover_name="StoreCountry",
    title="Revenue by country",
)

fig.update_layout(template="plotly_white")
fig.show()
```

```python
sales_by_month = (
    tx.groupby("month", as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"))
    .sort_values("month")
)

fig = px.line(sales_by_month, x="month", y="revenue",
              title="Monthly revenue trend")
fig.update_layout(template="plotly_white")
fig.show()
```

```python
sales_by_segment = (
    tx.groupby("ClientSegment", as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"),
         customers=("ClientID", "nunique"))
    .sort_values("revenue", ascending=False)
)

fig = px.bar(sales_by_segment, x="ClientSegment", y="revenue",
             title="Revenue by client segment")
fig.update_layout(template="plotly_white", xaxis_tickangle=-45)
fig.show()
```

```python
mix = (
    tx.groupby(["ClientSegment", "Category"], as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"))
)

fig = px.bar(mix, x="ClientSegment", y="revenue", color="Category",
             title="Category revenue mix by client segment",
             barmode="stack")
fig.update_layout(template="plotly_white", xaxis_tickangle=-45)
fig.show()
```

```python
mix_country = (
    tx.groupby(["StoreCountry", "Category"], as_index=False)
    .agg(revenue=("SalesNetAmountEuro", "sum"))
)

fig = px.bar(mix_country, x="StoreCountry", y="revenue", color="Category",
             title="Category revenue mix by store country",
             barmode="stack")
fig.update_layout(template="plotly_white", xaxis_tickangle=-45)
fig.show()
```
